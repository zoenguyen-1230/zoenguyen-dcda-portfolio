````python
df_latest_inspections = df.sort_values(by=['business_id', 'inspection_date'], ascending=[True, False]).drop_duplicates(subset=['business_id'])

# Explicitly drop rows with NaN inspection_score AFTER getting the latest inspections
df_latest_inspections.dropna(subset=['inspection_score'], inplace=True)

print("DataFrame with latest inspections. First 5 rows:")
print(df_latest_inspections.head())
print("\nDataFrame Info:")
print(df_latest_inspections.info())
````

````python
def categorize_score(score):
    if pd.isna(score):
        return 'Unknown'
    elif score >= 90:
        return 'Good'
    elif score >= 70:
        return 'Moderate'
    else:
        return 'Poor'

df_latest_inspections['score_category'] = df_latest_inspections['inspection_score'].apply(categorize_score)

print("DataFrame with score categories. First 5 rows with score_category:")
print(df_latest_inspections[['business_name', 'inspection_score', 'score_category']].head())
print("\nScore Category Value Counts:")
print(df_latest_inspections['score_category'].value_counts())
````

````python
def create_popup_html(row):
    business_name = row['business_name'] if pd.notna(row['business_name']) else 'Unknown'
    business_address = row['business_address'] if pd.notna(row['business_address']) else 'Unknown'

    inspection_score = row['inspection_score']
    # Since rows with NaN inspection_score are dropped, this will always be a number
    score_display = f"{int(inspection_score)} / 100"

    inspection_date = row['inspection_date']
    date_display = inspection_date.strftime('%Y-%m-%d') if pd.notna(inspection_date) else 'Unknown'

    risk_category = row['risk_category'] if pd.notna(row['risk_category']) and row['risk_category'] != 'Unknown' else 'Unknown'

    html = f"""
    <b>{business_name}</b><br>
    Address: {business_address}<br>
    Latest Score: {score_display}<br>
    Inspection Date: {date_display}<br>
    Risk Category: {risk_category}
    """
    return html

df_latest_inspections['popup_html'] = df_latest_inspections.apply(create_popup_html, axis=1)

print("First 5 rows with business_name and generated popup_html:")
print(df_latest_inspections[['business_name', 'popup_html']].head())
````

````python
import folium
import folium.plugins

# --- Configuration variables for map visualization --- #
MAX_MARKERS = 2000  # Limit the total number of markers on the map to reduce clutter
MARKER_RADIUS = 5   # Radius of the CircleMarkers
FILL_OPACITY = 0.7  # Transparency of marker fill
LINE_OPACITY = 1.0  # Transparency of marker outline
# --- End Configuration variables --- #

# 1. Determine the central coordinates for San Francisco
sf_coords = (37.7749, -122.4194) # San Francisco coordinates

# 2. Initialize a Folium map
m = folium.Map(location=sf_coords, zoom_start=12)

# 3. Define a function to assign marker colors based on score_category
def get_marker_color(score_category):
    if score_category == 'Good':
        return 'green'
    elif score_category == 'Moderate':
        return 'orange'
    elif score_category == 'Poor':
        return 'red'
    else:
        return 'gray' # For 'Unknown' scores

# Filter df_latest_inspections to MAX_MARKERS based on recency and score
# Sort by inspection_date (descending) to get most recent, then by inspection_score (ascending)
# to prioritize lower scores among equally recent inspections for early visibility.
df_display = df_latest_inspections.sort_values(by=['inspection_date', 'inspection_score'], ascending=[False, True]).head(MAX_MARKERS).copy()

# Create a MarkerCluster group for all restaurant markers (base layer)
# MarkerCluster helps reduce clutter by grouping nearby markers.
mc = folium.plugins.MarkerCluster(name='All Restaurants', options={'disableClusteringAtZoom': 15}).add_to(m)

# 4. Iterate through the filtered rows and add CircleMarkers
for idx, row in df_display.iterrows():
    if pd.notna(row['business_latitude']) and pd.notna(row['business_longitude']):
        color = get_marker_color(row['score_category'])
        folium.CircleMarker(
            location=[row['business_latitude'], row['business_longitude']],
            radius=MARKER_RADIUS,
            color=color,
            fill=True,
            fill_color=color,
            fill_opacity=FILL_OPACITY,
            opacity=LINE_OPACITY,
            popup=folium.Popup(row['popup_html'], max_width=300)
        ).add_to(mc)

# 5. Calculate the thresholds for 'Best Scores' and 'Worst Scores'
# This still uses the original df_latest_inspections to ensure calculation is on all data.
# Filter out NaN inspection scores for this calculation
df_scores_filtered = df_latest_inspections.dropna(subset=['inspection_score']).copy()

# Sort by score to find top and bottom 10%
df_scores_sorted = df_scores_filtered.sort_values(by='inspection_score', ascending=True)

# Calculate number of restaurants for top/bottom 10%
num_restaurants = len(df_scores_sorted)
num_top_bottom_10_percent = int(num_restaurants * 0.10)

# Get business_id for worst scores (bottom 10%)
worst_scores_ids = df_scores_sorted.head(num_top_bottom_10_percent)['business_id'].tolist()

# Get business_id for best scores (top 10%)
best_scores_ids = df_scores_sorted.tail(num_top_bottom_10_percent)['business_id'].tolist()

print("Folium map initialized with all restaurant markers.")
print(f"Number of restaurants displayed on map (MAX_MARKERS limit): {len(df_display)}")
print(f"Number of restaurants with 'Best Scores': {len(best_scores_ids)}")
print(f"Number of restaurants with 'Worst Scores': {len(worst_scores_ids)}")
````

````python
# folium.plugins is already imported in the previous cell

# 1. Filter for 'Poor' scores and prepare data for HeatMap
# The heatmap should still use all 'Poor' scores, not just the MAX_MARKERS subset.
df_poor_scores = df_latest_inspections[df_latest_inspections['score_category'] == 'Poor']
heat_data = df_poor_scores[['business_latitude', 'business_longitude']].values.tolist()

# 2. Create a Folium HeatMap layer
folium.plugins.HeatMap(heat_data, name='Poor Score Heatmap').add_to(m)

# 3. Create 'Best Scores' FeatureGroup and add CircleMarkers
best_scores_fg = folium.FeatureGroup(name='Best Scores').add_to(m)
# Filter for best scores that are also in the df_display (MAX_MARKERS) set.
df_best_display = df_display[df_display['business_id'].isin(best_scores_ids)]
for idx, row in df_best_display.iterrows():
    if pd.notna(row['business_latitude']) and pd.notna(row['business_longitude']):
        folium.CircleMarker(
            location=[row['business_latitude'], row['business_longitude']],
            radius=MARKER_RADIUS, # Use the same radius as base markers
            color='green',
            fill=True,
            fill_color='green',
            fill_opacity=FILL_OPACITY, # Use the same transparency as base markers
            opacity=LINE_OPACITY,
            popup=folium.Popup(row['popup_html'], max_width=300)
        ).add_to(best_scores_fg)

# 4. Create 'Worst Scores' FeatureGroup and add CircleMarkers
worst_scores_fg = folium.FeatureGroup(name='Worst Scores').add_to(m)
# Filter for worst scores that are also in the df_display (MAX_MARKERS) set.
df_worst_display = df_display[df_display['business_id'].isin(worst_scores_ids)]
for idx, row in df_worst_display.iterrows():
    if pd.notna(row['business_latitude']) and pd.notna(row['business_longitude']):
        folium.CircleMarker(
            location=[row['business_latitude'], row['business_longitude']],
            radius=MARKER_RADIUS, # Use the same radius as base markers
            color='red',
            fill=True,
            fill_color='red',
            fill_opacity=FILL_OPACITY, # Use the same transparency as base markers
            opacity=LINE_OPACITY,
            popup=folium.Popup(row['popup_html'], max_width=300)
        ).add_to(worst_scores_fg)

# 5. Add a LayerControl to the map
folium.LayerControl().add_to(m)

# 6. Display the map
m
````
